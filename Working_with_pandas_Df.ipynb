```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "Name" : ["Dj", "Rj", "Vp", "Rs", "Hk", "ab", "cd", "ef", "gh", "hd"],
    "Age" : np.random.randint(20,40, 10),
    "Sex" : np.random.choice(["Male", "Female"], 10),
    "Marks" : np.random.normal(40,10,10),
    "Salary" : np.random.normal(100000,5000,10).astype(int),
    "Weight" : np.random.randint(50,100,10)
})
```

```python
df["Bonus"] = df["Salary"]*0.2
df["After bonus"] = df["Salary"] + df["Bonus"]
```
