# Dataframe


*   2d
*   collecions of series
*   can store multiple types of data(Heterogeneous)



```python
df = pd.DataFrame({
    "Name" : ["Dj", "Rj", "Vp", "Rs", "Hk"],
    "Age" : [23,34,45,34,"23"],
    # "Branch" : ["cse", "cse", "ai"] #--> this will show the error because the size must be equal
    "Sex" : ["M", "M", "M", "F", "M"]
})
```

```python
df
```

**We can create dataframe in 2 ways... either by creating a dictionary or by using list of lists**
1.  using dictionary --> pd.DataFrame({"key" : [values]})
2.  using list of lists --> pd.DataFrame([ [row 1],
                                          [row 2] ], columns = [col names])

```python
df1 = pd.DataFrame([
    ["Dj", 23, "M"],
    ["Rj", 34, "M"],
    ["Vp", 45, "M"],
    ["Rs", 34, "F"],
    ["Hk", 23, "M"]],
    columns=["Name", "Age", "Sex"]
)
```

```python
df1
```

```python
df2 = pd.DataFrame({
    "Name" : ["Dj", "Rj", "Vp", "Rs", "Hk"],
    "Age" : np.random.randint(20,40, 5),
    "Sex" : ["M", "M", "M", "F", "M"],
    "Marks" : np.random.normal(40,10,5),
    "Salary" : np.random.normal(100000,5000,5).astype(int),
    "Weight" : np.random.randint(50,100,5)
})
```

```python
df2
```

```python
df2
```
