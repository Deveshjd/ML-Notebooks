```python
import numpy as mp
import pandas as pd
import seaborn as sns
```

```python
data = sns.load_dataset("tips")
data
```

```python
data["sex"].shape[0]
```

```python
# How many male came on thursday
data[(data["sex"] == "Male") & (data["day"] == "Thur")].shape[0]
```

```python
# Average tip given by all males
data[data["sex"] == "Male"]["tip"].mean()
```

```python
# No. of female smokers
data[(data["sex"] == "Female") & (data["smoker"] == "Yes")].shape[0]
```

```python
# Average size of female group
data[data["sex"] == "Female"]["size"].mean(). astype(int)
```

```python
# Average bill paid on sunday
data[data["day"] == "Sun"]["total_bill"].mean()
```

```python
# Highest bill ever paid for dinner
data[data["time"] == "Dinner"]["total_bill"].max()
```

```python
# Higest no. of smokers visited on which day
for i in data["day"].value_counts().index:
  print(f"{i} --> {data[(data["smoker"] == "Yes") & (data['day'] == i)].shape[0]}")
```

```python
# compare no. of females visited for lunch and dinner
for i in data["time"].value_counts().index:
  print(f"{i} --> {data[data["sex"] == "Female"].shape[0]}")
```

```python
# what time does the non smokers prefer(lunch, dinner)
for i in data["time"].value_counts().index:
  print(f"{i} --> {data[(data["smoker"] == "No") & (data["time"] == i)].shape[0]}")
```
