```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "Name" : ["Dj", "Rj", "Vp", "Rs", "Hk", "ab", "cd", "ef", "gh", "hd"],
    "Age" : np.random.randint(20,40, 10),
    "Sex" : np.random.choice(["Male", "Female"], 10),
    "Marks" : np.random.normal(40,10,10),
    "Salary" : np.random.normal(100000,5000,10).astype(int),
    "Height" : np.random.randint(50,100,10)
})
```

```python
import seaborn as sns
```

```python
data = sns.load_dataset("tips")
```

```python
data
```

```python
data["sex"]=data["sex"].map({"Male":0,"Female":1})
```

```python
data["smoker"]=data["smoker"].map({"Yes":1,"No":0})
```

```python
data["time"]=data["time"].map({"Dinner":0,"Lunch":1})
```

```python
data["day"]=data["day"].map({"Thur":1,"Fri":2, "Sat":3,"Sun":4})
```

```python
data["total_bill"]=data["total_bill"].round()
```
