# Scatter plot
**Used to show relation b/w 2 variables**

```python
x = [1,2,3,4,5,6,7,8,9]
y = [1,4,9,16,25,36,49,64,81]
x,y
```

```python
# Positive relationship
plt.figure()
plt.scatter(x,y)
plt.show()
```

```python
# Negative relationship
plt.figure()
plt.scatter(x,y[::-1])
plt.show()
```
