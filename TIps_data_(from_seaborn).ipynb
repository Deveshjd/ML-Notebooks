```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
data = sns.load_dataset("tips")
```

```python
data.head()
```

```python
data.groupby("day")["total_bill"].mean()
```

```python
data.groupby("day")[["total_bill", "tip", "size"]].mean()
```

```python
data.groupby("day")[["total_bill", "tip", "size"]].agg({
    "total_bill" : "mean",
    "tip" : "sum",
    "size" : ["count", "sum"]
})
```

# Pivot Table

```python
data.pivot_table(index = "day", values = "total_bill")
```

```python
data.pivot_table(index = "sex", values = "total_bill", columns="day")
```

```python
df = sns.load_dataset("flights")
```

```python
df
```

```python
df.pivot_table(index = "year", values = "passengers", columns = "month")
```

```python
df.pivot_table(index = "year", values = "passengers", columns = "month", margins=True)
```

```python
df.pivot_table(index = "year", values = "passengers", columns = "month", aggfunc = "sum", margins = True)
```

```python
df.pivot_table(index="year", values="passengers", aggfunc="min")
```
