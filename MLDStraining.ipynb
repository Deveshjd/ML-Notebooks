```python
# filter:
# used to filter out results
# ls = [1,2...15]
# ex. filter(lambda x = x>10, ls) ---> return no. greater than 10
```

```python
# filter the names starting with R or r
ls = ["Rohan", "Dj", "RIinku", "Suhani", "rinki"]
x = list(filter(lambda x : x[0]=='R' or x[0]=='r', ls))
x
```

```python
# use filter to extract even no. from list
ls = [i for i in range(1,20)]
x = list(filter(lambda x : x%2 == 0, ls))
x
```

```python
# use filter to extract odd no. from list
ls = [i for i in range(1,20)]
x = list(filter(lambda x : x%2 == 1, ls))
x
```

```python
# use filter to get no. >10 from a list
ls = [i for i in range(1,20)]
x = list(filter(lambda x : x>10, ls))
x
```

```python
# use filter() to filter out empty strings from a  list
ls = ["", "a", "b", ""]
x = list(filter(lambda x : x == "", ls))
x
```

```python
# use filter to keep palindromes from a list
ls = ["aba", "123", "121"]
x = list(filter(lambda x : x == x[::-1], ls))
x
```

# Day 3(Statistics)

---
**Statistics**

1.   Descriptive stats --> for big chunks of data --> Analysis of data
2.   Inferential stats --> for smaller data cunks --> hypothesis

Hypothesis -> 1. Null hypo, 2. Alternative Hypo

**Types of data:**
1. Qualitative data/ categorical (eg. gender, pincode)

    *1.1. Ordinal ---> ordered(eg. height)*

    *1.2. Nominal ---> no order or ranking(eg. blood group)*
2. Quantitative data/ Neumerical data

    *2.1. continuous data(decimal, measurable), eg. Age*

    *2.2. discrete data(countable), eg. numbers of students*
---

**Measure of central tendency** --> data ka centre kahan lie kar raha hai --> jo poore data ko represent kare

*   Mean --> Always goes towards majority  --> could be affected by outliers
*   Median --> if mean and meadian are same that means there is not such outliers in the data. The physical midpoint. It is often better than the mean when a dataset has extremely high or low outlier values.
*   Mode --> frequency


**Deviation :** mean(x) - x

**Standard Deviation :** squareRoot(sum(Deviation ^ 2))/ total pts.

-->ex. if mean is 60 and std dev is 10 => that means most of the data lies between 60-10 to 60+10 i.e (50, 70)

**Variance :** square of standard deviation, mostly used in machine learning

```python
from statistics import mean, mode, median
```

```python
age1 = [10, 20, 20, 20, 20, 20, 30, 21, 22]
age2 = [10, 20, 20, 20, 20, 20, 30, 21, 1000]
print("mean of age1 : ",mean(age1))
print("mean of age2 : ",mean(age2))
```

```python
print("median of age1 : ",median(age1))
print("median of age2 : ",median(age2))
```

```python
print("mode of age1 : ",mode(age1))
print("mode of age2 : ",mode(age2))
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# np.random.normal(x,y,z) ==> x: mean, y: standard deviation, z: no. of samples
list = np.random.normal(67, 10, 450)
```

```python
plt.plot(list, color='green', marker='o', linestyle='none')
plt.axhline(mean(list), color='red', label='Mean')
plt.axhline(median(list), color='blue', label='Median')
# plt.axhline(mode(list), color='yellow', label='Mode')
plt.axhline((np.mean(list)+np.std(list)), color='black', label='Standard Deviation')
plt.axhline((np.mean(list)-np.std(list)), color='black', label='Standard Deviation')
plt.legend()
plt.title('Normal Distribution')
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
height = np.random.normal(140, 10, 300)
```

```python
sns.histplot(height, bins = 30, kde=True)
plt.axvline(np.median(height), color='red', label='Median')
plt.axvline(np.mean(height), color='blue', label='Mean')
plt.axvline(np.mean(height) + np.std(height), color='black', label='Standard Deviation')
plt.axvline(np.mean(height) - np.std(height), color='black', label='Standard Deviation')
plt.legend()
plt.show()
```

**Normal distribution :** --> Bell shape curve

```python
# Skewed data
new_h = np.append(height, np.random.normal(500, 100, 10000))
sns.histplot(new_h, bins = 30, kde=True, color = 'blue')
plt.axvline(np.median(new_h), color='red', label='Median')
plt.axvline(np.mean(new_h), color='blue', label='Mean')
plt.axvline(np.mean(new_h) + np.std(height), color='black', label='Standard Deviation')
plt.axvline(np.mean(new_h) - np.std(height), color='black', label='Standard Deviation')
plt.legend()
plt.show()
```

```python
print(np.mean(new_h), np.median(new_h), sep = '\n')
```

```python
#  mean < median --> right skewness OR positive skewness
```

```python
# if mean > median ---> left skewness OR negative skewness
```
