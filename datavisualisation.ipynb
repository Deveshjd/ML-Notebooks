```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.array([10,20,30,25,50])
y = np.array([100,200,300,200,500])
```

```python
plt.figure() #plt.figure(dpi = 300) --> increses the sharpness of plot
plt.plot(x,y)
plt.show()
```

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(6,7), facecolor="lightblue")
plt.title("Line plot")
plt.xlabel("X-axis", color="green")
plt.ylabel("Y-axis", color="red")
plt.grid(True, color="red")
plt.plot(x,y, marker = "*", linestyle = "--", color = "orange")
plt.show()
```

```python
x1 = np.random.randint(10,50,10)
y1 = np.random.randint(10,50,10)
```

```python
print(x1,y1, sep="\n")
```

```python
plt.figure(facecolor = "lightgreen")
plt.grid(True, axis="x", color="red")
plt.title("Scatter Plot", fontstyle="italic", fontsize = 16)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.plot(x1,y1, linestyle="--", marker="s", mfc = "pink", mec = "green", ms=15, linewidth=4)
# mfc--> marker face color, mec--> marker edge color, me--> marker size
plt.tick_params(direction = "inout", length = 10, pad = 20)
plt.show()
```

```python
day = np.array(["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"])
temp = np.random.normal(20,5,7).astype(int)
```
