# **Starting with ML**

```python
data = pd.read_csv("/content/housing 1.csv")
```

```python
data.head()
```

```python
x = data.drop("median_house_value", axis=1)
```

```python
y = data["median_house_value"]
```

```python
cat_col_list = x.select_dtypes("object").columns.tolist()
num_col_list = x.select_dtypes(["int", "float"]).columns.tolist()
```

```python
print(cat_col_list)
print(num_col_list)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
num_pipeline = Pipeline(
    steps = [
        ("Imputer", SimpleImputer(strategy="median")),
        ("Scaler", StandardScaler())
    ]
)

num_pipeline
```

```python
cat_pipeline = Pipeline(
    steps = [
        ("Imputer", SimpleImputer(strategy='most_frequent')),
        ("Encoder", OneHotEncoder(sparse_output=False))
    ]
)

cat_pipeline
```

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("numerical", num_pipeline, num_col_list),
        ("categorical", cat_pipeline, cat_col_list)
    ]
)

preprocessor
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
x_train
```

```python
y_train
```

```python
x_train_scaled = preprocessor.fit_transform(x_train)
x_test_scaled = preprocessor.transform(x_test)
```

```python
# preprocessor.get_feature_names_out()
```

```python
# pd.DataFrame(x_train_scaled, columns=preprocessor.get_feature_names_out())
```

```python
x_test_scaled
```

```python
x_train_scaled
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, confusion_matrix, accuracy_score
```

```python
model = LinearRegression()
```

```python
model.fit(x_train_scaled, y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
r2_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True)
```
